### Point Estimation :
1. Mean(u)
   - 1 sample point estimation
   - multiple sample point estimation - (Central limit theorem)
2. Standard Deviation()

#### 1 sample point estimation

```python
sample_rs2=df.sample(frac=0.2) # randomly selecting 20% records
```

```python
# Let estimate about total bill
print('Sample mean :',sample_rs2['total_bill'].mean())
print('Sample std :',sample_rs2['total_bill'].std())
```

**Inference :**
- The avg bill amount in the restro is 22.3 USD.
- Bill amount can vary with 10.5 USD.

```python
# Let estimate about total bill
print('Population mean :',df['total_bill'].mean())
print('Population std :',df['total_bill'].std())
```

**OBS :**
- We can observe that one sample point estimation has enough gap in the sample vs population.
- To reduce this gap we can use multiple sample point estimation.

#### multiple sample point estimation

```python
sample_mean=[]
for i in range(100):
    df_sample1=df.sample(n=30)
    mean=df_sample1['total_bill'].mean()
    sample_mean.append(round(mean))
print('Sample Mean :',sample_mean)
```

```python
mean_series=pd.Series(sample_mean)
print("Average mean from all sample's mean:",mean_series.mean())
```

```python
print('Population Mean :',round(df['total_bill'].mean()))
```

### Central Limit Theorem
- If we take sufficient number of samples and calculate point estimates then it forms

```python
sns.histplot(mean_series,kde=True)
```

```python
# error
print('Simulated error',round(mean_series.std(),2))
```
