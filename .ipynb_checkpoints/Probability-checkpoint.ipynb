```python
#simulating 1000 events
li=[np.random.choice(['H','T']) for i in range(1000)]
li
```

```python
c=0
for i in li:
    if i=='H':
        c+=1
print('Probability of getting Head :',c/len(li))
```

```python
import pandas as pd
```

```python
df=pd.DataFrame({
    'Outcome':[np.random.choice(['H','T']) for i in range(1000)]
})
df
```

```python
# probability of getting head
fe=df[df['Outcome']=='H']
te=df
p=len(fe)/len(te)
print('Probability of getting Head :',p)
```

```python
df1=pd.DataFrame({
    'customer_id':list(range(100,1000)),
    'loan_amount':np.random.randint(100000,1000000,size=(900)),
    'status':[np.random.choice(['Fraud','Not Fraud']) for i in range(900)]
})
df1
```

```python
fe=df1[df1['status']=='Fraud']
te=df1
p=len(fe)/len(te)
print('Probability of getting next customer is Fraud is :',round(p,2))
```

### Axioms (Rules):
1. Probability cannot be negative.
2. Probability can only lie b/w 0 and 1.
3. Probability of all possible events is 1.
4. Complimentary probability = 1-p(A)
5. Additive Rule : one experiment multiple outcome.
   1. (Mutual Exclusive Event) : Does not effect each other. p(A or B) = P(A) + p(B)
   2. (Non Mutual Exclusive Event) : Does effect each other. p(A or B) = P(A) + p(B) - P(A n B)
6. Multiplicative Rule :
   1. (Independent Event) : One event cannot affect second means Probability of second event is same. P(A and B) = P(A) * P(B)
   2. (Dependent Event) :Reduce the Probability of second event. P(A and B) = P(A) * P(B/A)
