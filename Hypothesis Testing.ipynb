### Hypothesis :
- An assumption

1. Asssumption 1 : Avg age of M & F at innomatics is same.
2. Assumption 2 : M & F Proportion is different.
- This two are the conclusions of inferencial stats.

### Types of Hypothesis :
1. Null Hypothesis (H0) --> No change, established fact.
2. Alternate Hypothesis (H1) --> Change, Comparision...

**Note :** We have to test that the sample we have chosen provide us sufficient evidence to accept/regret H0.

**Rules :**
1. p_value < alpha --> Reject H0
2. p_value > alpha --> Accept H0

**Hypothesis Testing :**
1. Mean :
   + 1 sample with benchmark : ttest_1samp
   + 2 sample mean : ttest_ind
   + same sample before & after : ttest_rel
   + more than 2 sample mean : Anova
2. Proportion :
   + 1 proportion : 1 proportion 2-test
   + 2 proportion : 2 proportion 2-test
3. Categorical :
   + Chi-Square

#### Steps :
1. Define Hypothesis :H0 or H1
2. Consider alpha :0.05
3. Choose the statistical test
4. Compare p_value with alpha value : accept or Reject

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
df.head()
```

```python
sample = df.sample(frac=0.2)
sample.head()
```

```python
len(sample)
```

```python
print('Sample of mean :',round(sample['total_bill'].mean(),2))
```

```python
print('Population of mean :',round(df['total_bill'].mean(),2))
```

**Infer :** Based on sample mean, we can assume that avg total bill for all customers is 19 USD.

#### Hypothesis
- H0 : Population mean is 19.
- H1 : Population mean is not 19.

- Consider alpha value is 0.05
- Because we have 1 sample and benchmark 19:ttest_1samp

```python
from scipy.stats import ttest_1samp
```

```python
#method 1
result=ttest_1samp(sample['total_bill'],popmean=19.7)
```

```python
result.pvalue
```

```python
#method 2
t_stat, p_value = ttest_1samp(sample['total_bill'],popmean=19.7)
p_value
```

```python
p_value
```

```python
if p_value < 0.05:
    print('Reject H0')
else:
    print('Accept H1')
```

**Take two sample from the same data**
- H0 Mean of both sample is same
- H1 Mean of both sample is different

- Consider alpha : 0.05
  
**Ans :** We can use ttest_ind : Be are comparing two groups

```python
from scipy.stats import ttest_ind
```

```python
s1 = df.sample(frac=0.2)
s1.head()
```

```python
s2 = df.sample(frac=0.2)
s2.head()
```

```python
display(len(s1))
display(len(s2))
```

```python
print('Sample of mean :',round(s1['total_bill'].mean(),2))
print('Population of mean :',round(df['total_bill'].mean(),2))
```

```python
print('Sample of mean :',round(s2['total_bill'].mean(),2))
print('Population of mean :',round(df['total_bill'].mean(),2))
```

```python
#method 1
t_stat, p_value = ttest_ind(s1['total_bill'],s2['total_bill'])
if p_value < 0.05:
    print('Reject H0')
else:
    print('Accept H1')
```

### Error Types

### Hypothesis Test
1. parametric (Mean, Proportion) - Normal Data ---> More accurate result **EX :** ttest, Anova.. - Sample Age
2. Non Parametric (Rank, Ordered) - Skewed Data ---> Less accurate result **EX :** mamwhitneyu, Chi Square - Sample Ratios(1-5)

1. Parametric : from scipy.stats import ttest_ind
2. Non Parametric : from scipy.stats import mamwhitneyu
